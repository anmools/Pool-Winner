```python
import cv2
import numpy as np

I = cv2.imread('poolTable.png')

#convert to grayscale
G = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)

#find edges
E = cv2.Canny(G, threshold1=100, threshold2=200)

#find contours from edge image
contours, _ = cv2.findContours(
    E,
    mode=cv2.RETR_EXTERNAL,
    method=cv2.CHAIN_APPROX_NONE
)

#draw contours
cv2.drawContours(
    I,
    contours,
    contourIdx=-1,
    color=(0, 0, 255),
    thickness=2
)

cv2.imshow("Edges", E)
cv2.imshow("Contours", I)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
