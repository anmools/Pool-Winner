```python
import cv2
import numpy as np

I = cv2.imread('poolTable.png')

#convert to grayscale 
G = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)
GB = cv2.GaussianBlur(G, (5, 5), 0)

#find edges
E = cv2.Canny(GB, threshold1=100, threshold2=200)

#find contours from edge image
contours, _ = cv2.findContours(
    E,
    mode=cv2.RETR_EXTERNAL,
    method=cv2.CHAIN_APPROX_NONE
)

#check each contour
for c in contours:

    #find the length of each
    length = cv2.arcLength(c, closed=False)

    #ignore small contours
    if length > 30:

        #get width and height
        x, y, w, h = cv2.boundingRect(c)

        #compare
        ratio = w / h

        #print width, height and ratio
        print("w:", w, "h:", h, "ratio:", ratio)

        #check if the shape is circular
        if ratio > 0.7 and ratio < 1.3:

            if len(c) >= 5:

                #fit an ellipse to the contour
                ellipse = cv2.fitEllipse(c)

                cv2.ellipse(
                    I,
                    ellipse,
                    color=(0, 255, 0),
                    thickness=2
                )

cv2.imshow("Edges", E)
cv2.imshow("Detect ball shape", I)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
