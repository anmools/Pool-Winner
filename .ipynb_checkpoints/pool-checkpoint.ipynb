```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from matplotlib import image as image

I = cv2.imread('poolTable.png')

cv2.imshow("Pool Table", I)

cv2.waitKey(0)  
cv2.destroyAllWindows()
```

```python
# Turn the image to greyscale
G = cv2.cvtColor(I, cv2.COLOR_BGR2GRAY)

# Simple thresholding
#T, B = cv2.threshold(G, thresh = T, maxval = 255,type = cv2.THRESH_BINARY)

# Multiple Thresholds
RangeLower = ( 0, 0 , 150)
RangeUpper = ( 80 , 80 , 255)

B = cv2.inRange(I, RangeLower, RangeUpper)
# Use this to get the region of interest (pool ball)
ROI = cv2.bitwise_and(I,I, mask = B)

cv2.imshow("Mask", B)
cv2.imshow("ROI", ROI)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Converting to HSV. I have this so we can sepearate the colours and make them easier to pick out.
HSV = cv2.cvtColor(I,cv2.COLOR_BGR2HSV)
# This means the hue must be between the range 0-20, the saturation can be any saturation and the value means it's bright.
# The range was updated to fix the issue with the pale patches appearing. 
B = cv2.inRange(HSV, (0, 120, 150), (20, 255, 255))

# I got this from his notes but it looks a lot worse when I kept the parameters as (2,2)
shape = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

#Closing means it'll fill the holes inside the ball
B = cv2.morphologyEx(B, cv2.MORPH_CLOSE, shape)
#Opening will remove the stray pixels around it
B = cv2.morphologyEx(B, cv2.MORPH_OPEN, shape)

ROI = cv2.bitwise_and(I, I, mask=B)

cv2.imshow("Mask", B)
cv2.imshow("ROI", ROI)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
